### changing taper shape - important
varying the taper length slightly does shift the frequency, so instead of making the connection to the ring by making the taper longer, added a pad: now the frequency is 2.93 GHz \
changed from perfect conductor to copper, didn't seem to matter

making the connection shorter to 0.05mm, but keeping the taper shifts the frequency to 2.78 GHz

### expanding the ground plane
expanding the ground plane to 70 mm shifts the frequency to 2.81 GHz, at 60 mm it is 2.87 GHz \
changing the mesh to curvelinear changed the frequency and the peak field a bit, might also be which frequency we're sweeping \
changing the port size also seems to matter

### adding a solder mask
adding a solder mask like the one described in the paper shifts the frequency by 2 GHz; covering a smaller region shifts it less; adding a 0.01 mm air gap recovers the original frequency
