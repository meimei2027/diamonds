```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.optimize import curve_fit
import scipy.constants as cs
from scipy.interpolate import interp1d
```

#### Note
Hopefully I did this right

```python
na = 0.55
magnification = 30.
focal_tube_length = 160e-3
focal_length = focal_tube_length / magnification
solid_angle = np.arcsin(na) * 180 / cs.pi
```

#### Assuming that the laser obeys gaussian beam dynamics:

```python
# Define gamma to be some geometric factor which accounts for the beam shape changing from when we make the pulse. 
gamma = 1
beam_diameter = 4.8e-3
waist_final = focal_length * 532e-9/(cs.pi*beam_diameter/2*gamma)
print(f"waist_final = {waist_final*1e6:.2f} microns")
```

#### Assuming the beam penetrates the sample in a roughly cylindrical fashion:
ppm to density comes from this paper: https://www.sciencedirect.com/science/article/pii/S0925963523007975

```python
# Define d to be the penetration depth
nv_density = 6 # ppm 
nv_density_per_m3 = nv_density * 1.76e17 * 1e6 # ppm to cm^-3 to m^-3
d = 2e-6 # estimated depth from this paper in meters
volume = cs.pi * (waist_final)**2 * d
nvs_in_volume = nv_density_per_m3 * volume
print(f"nvs_in_volume = {nvs_in_volume}")
```

#### Assuming a pulse duration of 300ns, we can use this paper: https://arxiv.org/pdf/1511.04407.
To estimate the number of emitted photons in a pulse to be 0.06 photons per NV center

```python
photons_per_pulse = 0.06 * nvs_in_volume
# Define a sigma term to be the effects of photons being affected by interal reflections and scattering in the diamond
sigma = 0.18
print(f"photons_per_pulse = {photons_per_pulse*sigma:.2f}")
```

#### To calculate the number of photons going into the microscope aperature: 

```python
angular_radius = np.arctan(2 * na) * 180 / cs.pi # in terms of degrees
```

```python
total_coverage = angular_radius ** 2 / 57.2958 ** 2
print(f"number of photons through the aperature: {photons_per_pulse * total_coverage:.2f}")
```
