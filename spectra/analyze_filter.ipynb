```python
import numpy as np
import matplotlib.pyplot as plt

```

```python

def load_spectrum_txt(file_path, delimiter=","):
    data = np.loadtxt(file_path, delimiter=delimiter)

    wavelength_nm = data[:, 0]
    intensity_arb = data[:, 1]

    return wavelength_nm, intensity_arb


def plot_spectrum(ax, wavelength_nm, intensity, integration_time, label=""):
    ax.plot(wavelength_nm, intensity / (integration_time / 100e-3), label=label)

    ax.set_xlabel("Wavelength (nm)")
    ax.set_ylabel("Intensity (per 100ms)")

    ax.grid(True, alpha=0.3)

    return ax
```

```python
fig, ax = plt.subplots()
file_path = "filter_spectra/darkroom_laser_DMLP550.txt"
file_path2 = "filter_spectra/darkroom_white_light_DMLP550.txt"

wl, intensity = load_spectrum_txt(file_path)
wl2, intensity2 = load_spectrum_txt(file_path2)

ax = plot_spectrum(ax, wl, intensity, 100e-3)
ax = plot_spectrum(ax, wl2, intensity2, 1000e-3)


plt.legend()
plt.show()
```
